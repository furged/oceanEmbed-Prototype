```python
import copernicusmarine

info = copernicusmarine.describe(
    dataset_id="cmems_mod_glo_phy_my_0.083deg_P1D-m"
)

print(info)
```

```python
import copernicusmarine

copernicusmarine.subset(
    dataset_id="cmems_mod_glo_phy_my_0.083deg_P1D-m",
    variables=["thetao", "so", "zos"],
    minimum_longitude=75,
    maximum_longitude=76,
    minimum_latitude=10,
    maximum_latitude=11,
    minimum_depth=0,
    maximum_depth=10,
    start_datetime="2020-01-01",
    end_datetime="2020-01-07",
    output_directory="/content/oceanEmbed-Prototype/data",
    output_filename="test_subset.nc"
)
```

```python
import xarray as xr

ds = xr.open_dataset("/content/oceanEmbed-Prototype/data/test_subset.nc")

print(ds)
print("\nVariables:")
print(list(ds.data_vars))

print("\nDimensions:")
print(ds.dims)
```

```python
# Look at the actual numerical ranges of our variables

print("thetao:")
print(ds.thetao.min().item(), "to", ds.thetao.max().item())

print("\nso:")
print(ds.so.min().item(), "to", ds.so.max().item())

print("\nzos:")
print(ds.zos.min().item(), "to", ds.zos.max().item())
```

```python
# Show one temperature profile at one location and one time

profile = ds.thetao.isel(
    time=0,
    latitude=6,
    longitude=6
)

print(profile)
```

```python
# Extract the shallowest temperature and salinity as surface fields

sst = ds.thetao.isel(depth=0)
sss = ds.so.isel(depth=0)
ssh = ds.zos

print("SST shape:", sst.shape)
print("SSS shape:", sss.shape)
print("SSH shape:", ssh.shape)

print("\nSST:")
print(sst)

print("\nSSS:")
print(sss)

print("\nSSH:")
print(ssh)
```

```python
import numpy as np

# Extract surface variables
sst = ds.thetao.isel(depth=0)
sss = ds.so.isel(depth=0)
ssh = ds.zos

# Stack the 3 surface variables into input channels
X = np.stack([
    sst.values,
    sss.values,
    ssh.values
], axis=1)

print("X shape:", X.shape)
print("Expected: (time, channels, latitude, longitude)")

print("\nX min:", np.nanmin(X))
print("X max:", np.nanmax(X))
print("Number of NaNs:", np.isnan(X).sum())
```

```python
# Build the target: subsurface temperature

Y = ds.thetao.values

print("Y shape:", Y.shape)
print("Number of NaNs in Y:", np.isnan(Y).sum())

print("\nFinal test shapes:")
print("X:", X.shape, "  -> surface inputs")
print("Y:", Y.shape, "  -> temperature target")
```
